```python
from DT import DecisionTree
import numpy as np

class RandomForest:
    def __init__(self, n_estimators=100, max_depth=100, min_samples_split=2, n_feats=None):
        self.n_estimators = n_estimators
        self.max_depth = max_depth
        self.min_samples_split = min_samples_split
        self.n_feats = n_feats
        self.trees = []

    def fit(self, X, y):
        self.trees = []
        n_features = self.n_feats or max(1, int(np.sqrt(X.shape[1])))
        for _ in range(self.n_estimators):
            idxs = np.random.choice(len(X), len(X), replace=True)   # bootstrap
            tree = DecisionTree(max_depth=self.max_depth,
                                min_samples_split=self.min_samples_split,
                                n_features=n_features)
            tree.fit(X[idxs], y[idxs])
            self.trees.append(tree)

    def predict(self, X):
        preds = np.array([tree.predict(X) for tree in self.trees])  # (n_trees, n_samples)
        return np.array([np.bincount(col.astype(int)).argmax() for col in preds.T])
```

```python
from sklearn import datasets
from sklearn.model_selection import train_test_split

data = datasets.load_breast_cancer()
X_train, X_test, y_train, y_test = train_test_split(data.data, data.target, test_size=0.3, random_state=42)

rf = RandomForest(n_estimators=10, max_depth=5)
rf.fit(X_train, y_train)
print("Accuracy:", np.mean(rf.predict(X_test) == y_test))
```
